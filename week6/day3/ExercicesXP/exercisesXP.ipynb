🌟 Exercise 3 : Extracting Headers from Wikipedia’s Main Page

```python
from urllib.request import urlopen
from bs4 import BeautifulSoup

url = 'https://en.wikipedia.org/wiki/Main_Page'

response = urlopen(url)
html_content = response.read()

# Parse the HTML
soup = BeautifulSoup(html_content, 'html.parser')

# Extract all heading tags (h1, h2, h3, h4, h5, h6)
headers = soup.find_all(['h1', 'h2', 'h3', 'h4', 'h5', 'h6'])

# Display the extracted headers
for header in headers:
    print(header.name, ":", header.text)  # Print the tag name and text
```

🌟 Exercise 4 : Checking for Page Title

```python
from urllib.request import urlopen
from bs4 import BeautifulSoup

# Function to check if a page has a title
def check_page_title(url):
  try:
    # Open the URL and read the HTML content
    response = urlopen(url)
    html_content = response.read()

    # Parse the HTML
    soup = BeautifulSoup(html_content, 'html.parser')

    # Find the title tag
    title_tag = soup.title

    # Check if a title exists and print the result
    if title_tag and title_tag.text.strip():
      print(f"The page has a title: {title_tag.text}")
    else:
      print("The page does not have a title.")

  except Exception as e:
        print(f"Error fetching the page: {e}")

# Example usage
url = 'https://en.wikipedia.org/wiki/Main_Page'
check_page_title(url)
```

🌟 Exercise 5 : Analyzing US-CERT Security Alerts

```python
import requests
from bs4 import BeautifulSoup

# Base URL for CISA alerts
base_url = "https://www.cisa.gov/news-events/cybersecurity-advisories"
# filters
params = {
    "f[0]": "advisory_type:93",
    "f[1]": "release_date_year:2025",
    "page": 0  # Start at page 0
}

# Avoid being indentify as a bot :
# headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"}

total_alerts = 0

while True:
    # Send a GET request to the URL with pagination parameters
    # response = requests.get(base_url, params=params, headers=headers)
    response = requests.get(base_url, params=params)

    if response.status_code != 200:
        print(f"Failed to retrieve data. HTTP Status Code: {response.status_code}")
        break

    soup = BeautifulSoup(response.text, 'html.parser')

    # Find all elements that represent individual alerts
    alerts = soup.find_all('div', class_='c-view__row')

    # If no alert were found, stop the loop (end of pages)
    if not alerts:
        break

    # Add number of alertes found to the total
    total_alerts += len(alerts)
    print(f"Page {params['page']} : {len(alerts)} alerts founded")

    # Next page
    params["page"] += 1

print(f"Total number of security alerts issued by US-CERT in 2025: {total_alerts}")
```

🌟 Exercise 6 : Scraping Movie Details

```python
import requests
from bs4 import BeautifulSoup
import random

url = 'https://www.imdb.com/list/ls091294718/'

# Send a GET request to the IMDb page
# Avoid being indentify as a bot
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)",
           "Accept-Language": "en-US,en;q=0.9"}

response = requests.get(url, headers=headers)

if response.status_code != 200:
    print(f"Failed to retrieve data. HTTP Status Code: {response.status_code}")
else:
    # Parse the HTML content
    soup = BeautifulSoup(response.text, 'html.parser')
    # Find all movies in the list
    films = soup.find_all('li', class_='ipc-metadata-list-summary-item')

    # Shuffle and select 10 random movies
    random.shuffle(films) # shuffle the list randomly
    selected_movies = films[:10] # select the first 10 movies

    for index, film in enumerate(selected_movies, start=1):
      title = film.find('div', class_='ipc-title ipc-title--base ipc-title--title ipc-title-link-no-icon ipc-title--on-textPrimary sc-3713cfda-2 fSzZES dli-title with-margin').text.strip()
      year = film .find('span', class_='sc-f30335b4-7 jhjEEd dli-title-metadata-item').text.strip()
      summary = film.find('div', class_= 'ipc-html-content ipc-html-content--base sc-d49a611d-0 gKxuCN sttd-plot-container').text.strip()

      print(f"{index}. {title} ({year})")
      print(f"  Summary: {summary}\n")
```
