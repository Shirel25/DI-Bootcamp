```python
from openpyxl import Workbook
import pandas as pd
import os
from google.colab import files

# Creating a new Excel workbook and sheet
wb = Workbook()
ws = wb.active

# Add data to the sheet
ws["A1"] = "First number ==>"
ws["B1"] = 2

ws["A2"] = "Second number ==>"
ws["B2"] = 8

# Add the formula for multiplication in cell B3
ws["B3"] = "=B1*B2"

# Save the workbook to a file
wb.save("simple_calculator.xlsx")

# Download the file
files.download("simple_calculator.xlsx")

# Close the workbook
wb.close()
```
