```python
from openpyxl import Workbook

# Create a new workbook
wb = Workbook()

# Get the active worksheet and rename it
ws1 = wb.active
ws1.title = "Grades"
```

```python
data = {
    "Joe": {
        "math": 65,
        "science": 78,
        "english": 98,
        "gym": 89
    },
    "Bill": {
        "math": 55,
        "science": 72,
        "english": 87,
        "gym": 95
    },
    "Tim": {
        "math": 100,
        "science": 45,
        "english": 75,
        "gym": 92
    },
    "Sally": {
        "math": 30,
        "science": 25,
        "english": 45,
        "gym": 100
    },
    "Jane": {
        "math": 100,
        "science": 100,
        "english": 100,
        "gym": 60
    }
}
```

```python
import pandas as pd
from openpyxl.utils.dataframe import dataframe_to_rows

# Create a DataFrame from the dictionary
df = pd.DataFrame.from_dict(data, orient='index').reset_index()
df.rename(columns={'index': 'Student'}, inplace=True)

# Insert DataFrame into "Grades"
for row in dataframe_to_rows(df, index=False, header=True):
    ws1.append(row)
```

```python
# Add a formula in the last row of each subject column to calculate the average grade for that subject
num_students = len(df)

average_row_index = num_students + 2  # +1 for the header, +1 for an empty line
ws1[f"A{average_row_index}"] = "Average"

# Columns B to E = math, science, english, gym
subjects = ["B", "C", "D", "E"]

for col in subjects:
    # Example : B2:B6 if 5 students
    cell_range = f"{col}2:{col}{num_students+1}"
    formula_cell = f"{col}{average_row_index}"
    ws1[formula_cell] = f"=AVERAGE({cell_range})"
```

```python
from openpyxl.styles import Font, PatternFill

# Apply bold and colored formatting to the header row
header_row = ws1[1]

bold_font = Font(bold=True)
light_blue_fill = PatternFill(start_color="FFCCE5FF", end_color="FFCCE5FF", fill_type="solid")

# Apply filters
for cell in header_row:
    cell.font = bold_font
    cell.fill = light_blue_fill
```

```python
# Save the file
wb.save("grades_with_averages.xlsx")
```
