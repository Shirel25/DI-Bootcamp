🌟 **Exercise 1** : Defining the Problem and Data Collection for Loan Default Prediction

```python
# Loan default prediction

# Problem statement
probleme_statement = """
Loan default prediction is a critical challenge in the finacial sector.
Financial institutions, such as banks and credit unions, need to assess the risk
associated with lending money to individuals or businesses.
A loan default occurs when a borrow fails to make required payments, leading to finacial
losses for the lender.
The objective of this project is to develop a machine learning model that can predict whether an applicant is likely to default
on their loan. By accurately predicting defaults, financial institutions can mitigate risks, improve decision-making processes,
and offer tailored financial solutions.
"""

# Types of data
data_types = ["personal details of applicants", "credit scores", "loan amounts", "repayment history"]
data_sources = ["Financial Institution’s Internal Records", "Credit Bureaus and Rating Agencies"]

print("Probleme Statement : ", probleme_statement)
print("Data Types : ", data_types)
print("Data Sources : ", data_sources)
```

🌟 **Exercise 2** : Feature Selection and Model Choice for Loan Default Prediction

```python
# Feature Selection
selected_features = ['Education', 'Self_Employed', 'ApplicantIncome', 'CoapplicantIncome', 'LoanAmount', 'Loan_Amount_Term', 'Credit_History',
                     'Property_Area', 'Loan_Status']


# Justification for Feature Selection
feature_justification = """
- Education : Higher education levels may be correlated with better financial literacy and income stability,
impacting loan repayment ability.
- Self_Employed : Self-employed individuals may have more variable income streams compared to salaried applicants,
potentially increasing the risk of default.
- ApplicantIncome : Higher income generally suggests a lower risk of default.
- CoapplicantIncome : If a loan has a co-applicant, their income contributes to the total household income,
reducing the financial burden on the primary borrower.
- LoanAmount : Higher loan amounts may be riskier, especially if they are disproportionately large
compared to the applicant’s income.
- Loan_Amount_Term : A shorter-term loan means higher monthly payments, which might lead to defaults if income is insufficient.
- Credit_History : A strong credit history indicates timely repayments in the past, making it a strong predictor of future repayment behavior.
- Property_Area : The loan risk might vary depending on whether the property is in an urban, semi-urban, or rural area, as economic conditions
and employment opportunities differ across locations.
- Loan_Status : This is the outcome we are trying to predict (whether the loan was approved/repaid or defaulted).
"""

print(feature_justification)
```

🌟 **Exercise 5** : Designing an Evaluation Strategy for Different ML Models

```python
# Supervised Learning --> Random Forest (classification model)
supervised_evaluation = """
Metrics :
- accuracy : correct predictions / total predictions
- precision : correct positive predictions / total positive predictions
- recall : correct positive predictions / total actual positive instances
- F1-score : harmonic mean of precision and recall
- ROC curves : how well the model distinguishes between classes

Evaluations :
- Train-Test evaluation
- Cross-validation : helps assess model performance across different data split
- Confusion matrix : shows true positives, true negatives, false positives, false negatives

Challenges and Limitations :
- Overfitting
- Imbalanced datasets may lead to misleading accuracy scores
- Hyperparameter tuning is crucial for model performance
"""


# Unsupervised Learning --> K-means Clustering
unsupervised_evaluation = """
How to Assess Effectiveness :
- Silhouette Score : measures how well clusters are separated; values close
to 1 indicate well-defined clusters.
- Elbow Method : Plots the sum of squared distances vs. the number of clusters
to optimal number of clusters.
- Davies-Bouldin Index : Evaluates how compact and well-separated clusters are.

Challenges and Limitations :
- Chossing the right number of clusters is tricky.
- K-Means assumes spherical clusters, which may not always be the case.
- Sensitive to outliers ans scale of data.
"""


# Reinforcement Learning --> Q-Learning
reinforcement_evaluation = """
How to Measure Success :
- Cumulative Reward : Measures the total reward collected over time;
higher cumulative rewards indicate better performance.
- Convergence : Checks if the agent's Q-values stabilize over multiple episodes.
- Exploration vs. Exploitation Balance : Ensures the agent explores enough before
exploiting learned knowledge.

Challenges and Limitations :
- Learning rate tuning : A high learning rate may cause instability, while a low
one may lead to slow learning.
- Large state spaces make Q-tables impratical.
- Training time can be long for complex environments.
"""

print("Supervised evaluation :", supervised_evaluation)
print("-----------------------------------------------")
print("Unsupervised Learning : ", unsupervised_evaluation)
print("-----------------------------------------------")
print("Reinforcement Learning : ", reinforcement_evaluation)
```
